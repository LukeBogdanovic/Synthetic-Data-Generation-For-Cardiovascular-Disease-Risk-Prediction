```python
import torch
import matplotlib.pyplot as plt

%matplotlib inline
plt.rcParams.update({"font.size":16})
```

## Models withour MVDTW Term Augmenting Loss

### BiLSTM-CNN Model W/O MVDTW Term Augmenting Loss

```python
model = torch.load("models/BiLSTM_CNN_WGAN/Model_0_GP_10.0_DTW_0.0/Model.pth", weights_only=False)
metrics_history_1 = model['metrics_history']
print(f"MMD: {metrics_history_1['mmd'][-1]:.4f}")
print(f"MVDTW: {metrics_history_1['mvdtw'][-1]:.4f}")
print(f"W-Gap: {metrics_history_1['wgap'][-1]:.4f}")
```

```python
plt.plot(metrics_history_1['epoch'], metrics_history_1['mmd'])
plt.xlabel('Epoch')
plt.ylabel('MMD')
plt.title('MMD Change over Training Epochs for BiLSTM-CNN w/o MVDTW Loss Term')
plt.grid()
plt.show()
```

```python
plt.plot(metrics_history_1['epoch'], metrics_history_1['mvdtw'])
plt.xlabel('Epoch')
plt.ylabel('MVDTW')
plt.title('MVDTW Change over Training Epochs for BiLSTM-CNN w/o MVDTW Loss Term')
plt.grid()
plt.show()
```

```python
plt.plot(metrics_history_1['epoch'], metrics_history_1['wgap'])
plt.xlabel('Epoch')
plt.ylabel('Wasserstein Gap')
plt.title('W-Gap Change over Training Epochs for BiLSTM-CNN w/o MVDTW Loss Term')
plt.grid()
plt.show()
```

```python
plt.plot(metrics_history_1['epoch'], metrics_history_1['critic_loss'], label='Critic Loss')
plt.plot(metrics_history_1['epoch'], metrics_history_1['gen_loss'], label='Gen Loss')
plt.xlabel('Epoch')
plt.ylabel('Wasserstein Loss Score')
plt.title('Loss Changes over Training Epochs for BiLSTM-CNN w/o MVDTW Loss Term')
plt.grid()
plt.legend()
plt.show()
```
