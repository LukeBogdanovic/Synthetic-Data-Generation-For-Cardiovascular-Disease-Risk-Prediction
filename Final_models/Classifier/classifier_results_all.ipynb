```python
import sys
PROJECT_ROOT = r"C:\\Dev\\Synthetic-Data-Generation-For-Cardiovascular-Disease-Risk-Prediction\\"
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import torch
%matplotlib inline
plt.rcParams.update({"font.size":16})
```

## Synthetic Trained Classifier

```python
class_names = ['None', 'MI', 'Stroke', 'Syncope']
classifier_metrics = torch.load("models/classifier24/checkpoint.pth", weights_only=False)
class_history = classifier_metrics['history']
class_synth_metrics = classifier_metrics['synth_test']
class_real_metrics = classifier_metrics['real_test']
```

```python
class_synth_metrics
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(class_synth_metrics['conf_mat'], annot=True,fmt='d',cmap='Blues',xticklabels=class_names, yticklabels=class_names)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Fully Synthetic Dataset Model Confusion Matrix For Real Test")
plt.show()
```

```python
class_real_metrics
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(class_real_metrics['conf_mat'], annot=True,fmt='d',cmap='Blues',xticklabels=class_names, yticklabels=class_names)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Fully Synthetic Dataset Model Confusion Matrix For Real Test")
plt.show()
```
